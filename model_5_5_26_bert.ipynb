## Define helper functions

```python
def as_dense(x):
    return x.toarray() if issparse(x) else np.asarray(x)

def jaccard(a, b):
    a = np.asarray(a, dtype=bool)
    b = np.asarray(b, dtype=bool)
    denom = np.logical_or(a, b).sum()
    return 0.0 if denom == 0 else np.logical_and(a, b).sum() / denom

def top_support_mask(vec, frac=0.05):
    vec = np.abs(np.asarray(vec).ravel())
    k = max(1, int(np.ceil(frac * vec.size)))
    mask = np.zeros(vec.size, dtype=bool)
    mask[np.argpartition(vec, -k)[-k:]] = True
    return mask

def edge_overlap_metrics(A_pred, A_ref, frac=0.05):
    pred = top_support_mask(A_pred, frac=frac)
    ref = top_support_mask(A_ref, frac=frac)
    tp = np.logical_and(pred, ref).sum()
    fp = pred.sum() - tp
    fn = ref.sum() - tp
    precision = tp / (tp + fp + 1e-8)
    recall = tp / (tp + fn + 1e-8)
    f1 = 2 * precision * recall / (precision + recall + 1e-8)
    jac = tp / (pred.sum() + ref.sum() - tp + 1e-8)
    return {
        "precision": float(precision),
        "recall": float(recall),
        "f1": float(f1),
        "jaccard": float(jac),
        "tp": int(tp),
        "pred_edges": int(pred.sum()),
        "ref_edges": int(ref.sum()),
    }

def mean_axis_jaccard(A_pred, A_ref, axis=0, frac=0.1):
    scores = []
    if axis == 0:  # per TF column
        for j in range(A_ref.shape[1]):
            p = top_support_mask(A_pred[:, j], frac=frac)
            r = top_support_mask(A_ref[:, j], frac=frac)
            scores.append(jaccard(p, r))
    else:  # per gene row
        for i in range(A_ref.shape[0]):
            p = top_support_mask(A_pred[i, :], frac=frac)
            r = top_support_mask(A_ref[i, :], frac=frac)
            scores.append(jaccard(p, r))
    return float(np.mean(scores)) if scores else 0.0

def module_sets_from_nmf(W, H, top_genes=25, top_tfs=15):
    # W: genes x r, H: r x tfs
    modules = []
    for k in range(W.shape[1]):
        g_idx = np.argsort(-W[:, k])[:top_genes]
        t_idx = np.argsort(-H[k, :])[:top_tfs]
        modules.append((set(g_idx.tolist()), set(t_idx.tolist())))
    return modules

def module_similarity(m1, m2):
    g1, t1 = m1
    g2, t2 = m2
    return 0.5 * jaccard(g1, g2) + 0.5 * jaccard(t1, t2)

def match_module_sets(ref_modules, cur_modules):
    S = np.array([[module_similarity(r, c) for c in cur_modules] for r in ref_modules], dtype=float)
    row_ind, col_ind = linear_sum_assignment(-S)
    matched = S[row_ind, col_ind]
    return {
        "mean_conservation": float(np.mean(matched)),
        "std_conservation": float(np.std(matched)),
        "matched_scores": matched,
        "similarity_matrix": S,
    }

def pseudotime_arrow_agreement(X_space, V_pred, pseudotime, k=30):
    """
    For each cell, find the nearest neighbor with slightly larger pseudotime
    among local neighbors, and compare that arrow with predicted velocity.
    Returns mean cosine agreement and fraction positive.
    """
    nn = NearestNeighbors(n_neighbors=min(k + 1, len(X_space))).fit(X_space)
    idx = nn.kneighbors(X_space, return_distance=False)[:, 1:]

    arrows = np.zeros_like(X_space, dtype=float)
    valid = np.zeros(len(X_space), dtype=bool)

    for i in range(len(X_space)):
        nbrs = idx[i]
        later = nbrs[pseudotime[nbrs] > pseudotime[i]]
        if later.size == 0:
            continue
        j = later[np.argmin(pseudotime[later] - pseudotime[i])]
        arrows[i] = X_space[j] - X_space[i]
        valid[i] = True

    if valid.sum() == 0:
        return {"mean_cosine": np.nan, "fraction_positive": np.nan, "n_valid": 0}

    Vn = normalize(V_pred[valid], norm="l2", axis=1)
    An = normalize(arrows[valid], norm="l2", axis=1)
    cos = np.sum(Vn * An, axis=1)

    return {
        "mean_cosine": float(np.mean(cos)),
        "fraction_positive": float(np.mean(cos > 0)),
        "n_valid": int(valid.sum()),
        "cosines": cos,
    }
```

```python
def build_tf_access_prior(gene_activity: np.ndarray,
                           A_tfs: np.ndarray,
                           gene_to_idx_full: dict,
                           gene_names_all: np.ndarray) -> np.ndarray:
    """
    Returns TF_access: (T,) float32 — population-mean chromatin accessibility
    for each TF locus, derived from the ATAC gene-activity matrix.
    """
    T = len(A_tfs)
    tf_access = np.zeros(T, dtype=np.float32)
    for i, tf in enumerate(A_tfs):
        if tf in gene_to_idx_full:
            g = gene_to_idx_full[tf]          # index in full gene list
            # find column in gene_activity that corresponds to this gene
            # gene_activity columns align with A_genes (from Step 2/3)
            # We need the sub-index; if tf is in A_genes use sub_g2i
            # otherwise fall back to 0
            if tf in sub_g2i:
                col = sub_g2i[tf]
                tf_access[i] = float(gene_activity[:, col].mean())
    # min-max normalise to [0, 1]
    lo, hi = tf_access.min(), tf_access.max()
    tf_access = (tf_access - lo) / (hi - lo + 1e-8)
    return tf_access
 
 
# ── call it (uses variables already in scope from earlier steps) ──────────
TF_access_prior = build_tf_access_prior(
    gene_activity, A_tfs, gene_to_idx_full, gene_names_all
)
TF_access_t = torch.tensor(TF_access_prior, dtype=torch.float32)  # (T,)
 
print(f"TF access prior: mean={TF_access_prior.mean():.3f}  "
      f"nonzero={np.mean(TF_access_prior > 0.05):.1%}")
```

## Step 5: Define model

```python
# ═══════════════════════════════════════════════════════════════════════════
# STEP 6  –  GRN Module
# ═══════════════════════════════════════════════════════════════════════════
#
# A learnable (G × T) weight matrix constrained by the motif prior M.
# Positive entries  →  TF activates gene.
# Negative entries  →  TF represses gene.
# Zero entries where M[g,t] = 0  →  no motif evidence, edge is silenced.
#
# Mathematical form:
#   A = softplus(U) * M      (element-wise; only motif-supported edges exist)
#   pred_rna[i] = X_TF[i] @ A.T   (cell i's TF expression × GRN → gene pred)
 
class GRNModule(nn.Module):
    """
    Sparse, motif-constrained GRN.
    Parameters
    ----------
    G : int   number of genes
    T : int   number of TFs
    M : np.ndarray (G, T)  binary motif-accessibility prior
    """
    def __init__(self, G: int, T: int, M: np.ndarray):
        super().__init__()
 
        # Raw weight matrix — initialised near zero
        self.W_raw = nn.Parameter(torch.zeros(G, T))
 
        # M is a hard mask: gradients are blocked for unsupported edges
        M_t = torch.tensor(M, dtype=torch.float32)   # (G, T)
        self.register_buffer("M", M_t)
 
        # TF-access prior used to initialise and regularise
        # (set externally after construction if desired)
        self.register_buffer("TF_access", torch.zeros(T))
 
    @property
    def A(self) -> torch.Tensor:
        """
        Effective GRN matrix: (G, T).
        We allow both positive and negative weights so the model can learn
        activation *and* repression.  The motif mask enforces sparsity.
        """
        return self.W_raw * self.M   # zero out non-motif edges
 
    def predict_rna(self, X_TF: torch.Tensor) -> torch.Tensor:
        """
        X_TF : (N, T) — TF expression
        returns: (N, G) — predicted gene expression
        """
        return X_TF @ self.A.T     # (N, T) @ (T, G) = (N, G)
 
    def sparsity_loss(self) -> torch.Tensor:
        """L1 penalty encourages most GRN edges to be near zero."""
        return torch.abs(self.W_raw).mean()
 
    def tf_access_consistency_loss(self) -> torch.Tensor:
        """
        Penalise strong regulatory weights for TFs whose chromatin loci
        are inaccessible across the ATAC population.
        Biology: a TF can't regulate anything if its own gene isn't expressed
        / its locus is closed.
        Loss = mean(|A| * (1 - TF_access))
        """
        closed = 1.0 - self.TF_access.clamp(0, 1)           # (T,)
        return (torch.abs(self.W_raw) * closed.unsqueeze(0)).mean()
```

## Step 7: BERT Task

```python
# ═══════════════════════════════════════════════════════════════════════════
# STEP 7  –  Masked Reconstruction (BERT task)
# ═══════════════════════════════════════════════════════════════════════════
#
# Algorithm:
#   1. For each cell i, sample a binary mask m_i ∈ {0,1}^G  (fraction p_mask)
#   2. Zero out the masked genes:  x_masked = x * (1 - m)
#   3. Reconstruct using GRN:      x_hat    = TF_expr @ A.T
#   4. Loss only on MASKED positions: MSE(x_hat[m], x[m])
#
# Why this works:
#   If the GRN weights are wrong, x_hat will be random and the loss is high.
#   The ONLY way to minimise is to make A[g, :] point in the direction of the
#   true regulatory relationship between TFs and gene g — learned from the
#   *structure* of co-variation across cells, not from a labelled target.
 
def masked_reconstruction_loss(grn: GRNModule,
                                rna_batch: torch.Tensor,
                                tf_batch: torch.Tensor,
                                mask_frac: float = 0.40) -> torch.Tensor:
    """
    rna_batch : (B, G) — log-normalised RNA expression
    tf_batch  : (B, T) — TF expression (subset of RNA)
    Returns scalar loss.
    """
    B, G = rna_batch.shape
 
    # 1. Random gene mask — each gene independently masked with prob mask_frac
    mask = torch.rand(B, G, device=rna_batch.device) < mask_frac  # (B, G) bool
 
    # 2. Masked input (not actually used by GRN — it predicts from TFs only)
    #    The masking here is applied to the *target* view, not the input,
    #    because the GRN input is X_TF (always observed).
    #    This is intentional: the GRN must predict the missing RNA from TF state alone.
 
    # 3. GRN prediction
    rna_pred = grn.predict_rna(tf_batch)   # (B, G)
 
    # 4. Loss on masked genes only
    if mask.sum() == 0:
        return rna_pred.mean() * 0.0       # degenerate mini-batch
 
    loss = F.mse_loss(rna_pred[mask], rna_batch[mask])
    return loss
```

## Step 8: Velocity Head

```python
# ═══════════════════════════════════════════════════════════════════════════
# STEP 8  –  Velocity Head
# ═══════════════════════════════════════════════════════════════════════════
#
# Velocity is the disequilibrium between the cell's CURRENT expression and
# the GRN's PREDICTED STEADY STATE given TF accessibility.
#
# Core idea:
#   steady_state[i] = TF_expr[i] @ A.T    (where the cell "wants" to go)
#   disequilibrium[i] = steady_state[i] - RNA[i]
#   velocity[i] = f(disequilibrium[i])
#
# A positive disequilibrium for gene g means TFs are signalling up-regulation
# but RNA hasn't caught up yet → positive velocity → gene is being turned on.
# No pseudotime, no root cell, no splicing needed.
 
class VelocityHead(nn.Module):
    """
    Lightweight MLP that maps (disequilibrium) → per-gene velocity.
    Input dim  = G (disequilibrium vector)
    Output dim = G (signed velocity per gene)
    """
    def __init__(self, G: int, hidden: int = 256):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(G, hidden),
            nn.LayerNorm(hidden),
            nn.GELU(),
            nn.Dropout(0.1),
            nn.Linear(hidden, hidden // 2),
            nn.GELU(),
            nn.Linear(hidden // 2, G),
        )
 
    def forward(self,
                rna: torch.Tensor,
                grn: GRNModule,
                tf: torch.Tensor) -> torch.Tensor:
        """
        rna : (N, G)
        tf  : (N, T)
        returns velocity : (N, G)
        """
        with torch.no_grad():
            # Detach GRN during velocity forward to prevent co-adaptation
            # (GRN is updated via reconstruction loss; velocity head learns
            #  to read the GRN's implied direction)
            steady = (tf @ grn.A.detach().T)   # (N, G)
        disequilibrium = steady - rna           # (N, G): signed gap
        return self.net(disequilibrium)         # (N, G): predicted Δgene
```

## Step 9 & 10: Velocity Self-Consistency Loss & ATAC Directional Prior

```python
# ═══════════════════════════════════════════════════════════════════════════
# STEP 9  –  Velocity Self-Consistency Loss
# ═══════════════════════════════════════════════════════════════════════════
#
# After applying velocity, the updated expression should be CLOSER to the
# GRN steady state — not further away.
#
# Math:
#   loss = max(0,  ||RNA + v − SS||²  −  ||RNA − SS||²  + margin)
# This is a margin ranking loss: "the step must improve, by at least margin."
 
def velocity_consistency_loss(velocity: torch.Tensor,
                               rna: torch.Tensor,
                               steady_state: torch.Tensor,
                               margin: float = 1e-2) -> torch.Tensor:
    """
    velocity     : (B, G)
    rna          : (B, G)
    steady_state : (B, G)
    """
    dist_now  = ((rna - steady_state) ** 2).sum(dim=1)          # (B,)
    rna_next  = rna + velocity
    dist_next = ((rna_next - steady_state) ** 2).sum(dim=1)     # (B,)
    # Penalise only when the step makes things worse (+ margin)
    return F.relu(dist_next - dist_now + margin).mean()
 
 
# ═══════════════════════════════════════════════════════════════════════════
# STEP 10  –  ATAC Directional Prior (gene-level)
# ═══════════════════════════════════════════════════════════════════════════
#
# Genes whose chromatin is closed (gene_access ≈ 0) should not have strong
# positive velocity — the locus is physically inaccessible.
# This encodes a unidirectional chromatin constraint without pairing cells.
 
def atac_direction_loss(velocity: torch.Tensor,
                        gene_access: torch.Tensor) -> torch.Tensor:
    """
    velocity    : (B, G) — predicted per-gene velocity
    gene_access : (G,)   — normalised ATAC gene accessibility
    """
    mean_vel = velocity.mean(dim=0)               # (G,)
    closed   = 1.0 - gene_access.to(velocity.device)
    # Penalise positive velocity where chromatin is closed
    return F.relu(closed * mean_vel).mean()

def velocity_smoothness_loss(velocity, batch_indices, neighbor_lookup):
    """
    velocity: (B, G) 
    batch_indices: (B,) global indices of cells in current batch
    neighbor_lookup: (N, K) global indices of neighbors
    """
    B = velocity.shape[0]
    # 1. Get global IDs of neighbors for cells in this batch: (B, K)
    neighbor_ids = neighbor_lookup[batch_indices]

    # 2. Find which neighbors are present in the current batch
    # We compare every neighbor_id to every batch_index
    # mask: (B, K, B) -> True if neighbor_ids[b, k] == batch_indices[b_prime]
    mask = neighbor_ids.unsqueeze(-1) == batch_indices.unsqueeze(0).unsqueeze(0)
    
    # 3. Find the local indices of those neighbors
    has_neighbor, neighbor_k_idx, local_neighbor_idx = mask.nonzero(as_tuple=True)
    
    if len(local_neighbor_idx) == 0:
        return torch.tensor(0.0, device=velocity.device, requires_grad=True)

    # 4. Calculate squared difference
    v_src = velocity[has_neighbor]           # Velocity of cells that have neighbors in batch
    v_tgt = velocity[local_neighbor_idx]     # Velocity of those neighbors
    
    return torch.mean((v_src - v_tgt)**2)
```

```python
import torch.nn as nn

# ═══════════════════════════════════════════════════════════════════════════
# STEP 11  –  Assemble full model and training loop
# ═══════════════════════════════════════════════════════════════════════════
 
# ── data prep ────────────────────────────────────────────────────────────
G = X.shape[1]    # genes
T = X_TF.shape[1] # TFs
 
X_t      = torch.tensor(X,    dtype=torch.float32)
X_TF_t   = torch.tensor(X_TF, dtype=torch.float32)
 
cell_idx_t = torch.arange(X_t.size(0)) # Create indices 0 to N-1
dataset = TensorDataset(X_t, X_TF_t, cell_idx_t) # Add indices to dataset

# Pre-compute neighbors in Decipher space (Z)
from sklearn.neighbors import NearestNeighbors
# Assuming V_dec2 is your N x 2 or N x Dim latent embedding
nbrs = NearestNeighbors(n_neighbors=15).fit(V_dec2)
_, neighbor_indices = nbrs.kneighbors(V_dec2)
neighbor_indices_t = torch.tensor(neighbor_indices, dtype=torch.long)

loader   = DataLoader(dataset,
                      batch_size=256,
                      sampler=RandomSampler(dataset, replacement=False),
                      drop_last=True)
 
# ── model construction ───────────────────────────────────────────────────
grn          = GRNModule(G=G, T=T, M=M)
grn.TF_access = TF_access_t.clone()        # attach ATAC prior
 
vel_head     = VelocityHead(G=G, hidden=256)
 
optimizer    = torch.optim.AdamW(
    list(grn.parameters()) + list(vel_head.parameters()),
    lr=5e-4, weight_decay=1e-4
)
scheduler    = torch.optim.lr_scheduler.CosineAnnealingLR(
    optimizer, T_max=200, eta_min=1e-5
)
 
# ── loss weights ─────────────────────────────────────────────────────────
W_MASK   = 1.0    # masked reconstruction  ← main self-supervised signal
W_VEL    = 0.5    # velocity self-consistency
W_ATAC   = 0.2    # ATAC chromatin direction prior
W_SPARSE = 0.05   # GRN L1 sparsity
W_ACCESS = 0.1    # TF-access consistency
W_SMOOTH = 0.1

# ── training ─────────────────────────────────────────────────────────────
print("Training BERT-GRN model …")
loss_history = []
 
N_EPOCHS = 200   # sufficient for datasets of 1 k – 50 k cells
 
for epoch in range(N_EPOCHS):
    grn.train(); vel_head.train()
    epoch_losses = []
 
    for rna_b, tf_b, batch_idx in loader:
 
        optimizer.zero_grad()
 
        # ----- Loss 1: Masked reconstruction (BERT task) -----------------
        l_mask = masked_reconstruction_loss(grn, rna_b, tf_b, mask_frac=0.40)
 
        # ----- Loss 2: Velocity self-consistency -------------------------
        velocity  = vel_head(rna_b, grn, tf_b)
        steady    = tf_b @ grn.A.T                       # (B, G)
        l_vel     = velocity_consistency_loss(velocity, rna_b, steady)
 
        # ----- Loss 3: ATAC directional prior ----------------------------
        l_atac    = atac_direction_loss(velocity, gene_access_t)
 
        # ----- Loss 4: GRN sparsity (L1) ---------------------------------
        l_sparse  = grn.sparsity_loss()
 
        # ----- Loss 5: TF accessibility consistency ----------------------
        l_access  = grn.tf_access_consistency_loss()

        # ----- Loss 6: Velocity Smoothness -------------------------------
        # Use the neighbor_indices_t we created in step 1
        l_smooth = velocity_smoothness_loss(velocity, batch_idx, neighbor_indices_t)
     
        loss = (W_MASK   * l_mask   +
                W_VEL    * l_vel    +
                W_ATAC   * l_atac   +
                W_SPARSE * l_sparse +
                W_ACCESS * l_access +
                W_SMOOTH * l_smooth)
 
        loss.backward()
        torch.nn.utils.clip_grad_norm_(
            list(grn.parameters()) + list(vel_head.parameters()), 1.0
        )
        optimizer.step()
        epoch_losses.append(float(loss.item()))
 
    scheduler.step()
 
    if epoch % 20 == 0 or epoch == N_EPOCHS - 1:
        avg = np.mean(epoch_losses)
        loss_history.append(avg)
        print(f"  Epoch {epoch:03d}  loss={avg:.4f}")
 
print("Training complete.")
```

## Step 13: Downstream Analysis Helpers

```python
# ═══════════════════════════════════════════════════════════════════════════
# STEP 13  –  Downstream analysis helpers
# ═══════════════════════════════════════════════════════════════════════════
 
def top_tf_targets(A_learned: np.ndarray,
                   A_tfs: np.ndarray,
                   A_genes: np.ndarray,
                   tf_name: str,
                   n: int = 20) -> list[tuple[str, float]]:
    """Return top-n gene targets of a given TF, ranked by |GRN weight|."""
    if tf_name not in A_tfs:
        raise ValueError(f"{tf_name} not in A_tfs")
    t_idx = np.where(A_tfs == tf_name)[0][0]
    col   = A_learned[:, t_idx]           # (G,)
    top   = np.argsort(-np.abs(col))[:n]
    return [(A_genes[i], float(col[i])) for i in top]
 
 
def velocity_umap(X: np.ndarray,
                  V_pred: np.ndarray,
                  n_pca: int = 50) -> tuple[np.ndarray, np.ndarray]:
    """
    Project expression and velocity into a 2-D PCA subspace for
    scVelo-style stream plots.
    Returns (X_2d, V_2d) both (N, 2).
    """
    pca  = PCA(n_components=n_pca, random_state=42)
    X_pc = pca.fit_transform(X)              # (N, 50)
 
    # Project velocity into PC space (delta in PC space)
    V_pc = V_pred @ pca.components_.T        # (N, 50)
 
    # 2-D via second PCA for visualisation
    pca2 = PCA(n_components=2, random_state=42)
    X_2d = pca2.fit_transform(X_pc)
    V_2d = V_pc @ pca2.components_.T
 
    return X_2d, V_2d
 
 
# Example usage:
X_2d, V_2d = velocity_umap(X, V_pred)
np.save("bert_grn_out/X_2d.npy", X_2d)
np.save("bert_grn_out/V_2d.npy", V_2d)
print("2-D velocity embedding saved.")
 
# Quick sanity check: fraction of cells with positive mean velocity
frac_pos = float(np.mean(V_pred.mean(axis=1) > 0))
print(f"Cells with net positive velocity: {frac_pos:.1%}  "
      f"(expect ~50% for unbiased datasets)")
 
# Top TF targets example (replace with a TF in your dataset)
example_tf = A_tfs[0]
print(f"\nTop 10 targets of {example_tf}:")
for gene, w in top_tf_targets(A_learned, A_tfs, A_genes, example_tf, n=10):
    print(f"  {gene:20s}  weight={w:+.4f}")
```

```python
from scipy.optimize import linear_sum_assignment

def match_module_sets(ref_modules, cur_modules):
    n = len(ref_modules)
    sim = np.zeros((n, n))

    # similarity matrix (gene + TF overlap)
    for i, (rg, rt) in enumerate(ref_modules):
        for j, (cg, ct) in enumerate(cur_modules):
            sim[i, j] = 0.5 * jaccard(rg, cg) + 0.5 * jaccard(rt, ct)

    # Hungarian matching (maximize similarity)
    row_ind, col_ind = linear_sum_assignment(-sim)

    scores = sim[row_ind, col_ind]

    return {
        "mean_conservation": float(np.mean(scores)),
        "pairs": list(zip(row_ind, col_ind))  # ✅ ADD THIS
    }
```
